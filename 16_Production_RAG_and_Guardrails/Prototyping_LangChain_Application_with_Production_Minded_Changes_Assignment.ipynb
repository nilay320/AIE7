#### Production Caching Architecture

Our LLMOps library implements sophisticated caching at multiple levels:

**Embedding Caching:**
The process of embedding is typically very time consuming and expensive:

1. Send text to OpenAI API endpoint
2. Wait for processing  
3. Receive response
4. Pay for API call

This occurs *every single time* a document gets converted into a vector representation.

**Our Caching Solution:**
1. Check local cache for previously computed embeddings
2. If found: Return cached vector (instant, free)
3. If not found: Call OpenAI API, store result in cache
4. Return vector representation

**LLM Response Caching:**
Similarly, we cache LLM responses to avoid redundant API calls for identical prompts.

**Benefits:**
- ⚡ Faster response times (cache hits are instant)
- 💰 Reduced API costs (no duplicate calls)  
- 🔄 Consistent results for identical inputs
- 📈 Better scalability

Our ProductionRAGChain automatically handles all this caching behind the scenes!

What are some limitations you can see with this caching approach? When is this most/least useful for production systems? 

Consider:
- **Memory vs Disk caching trade-offs**
- **Cache invalidation strategies** 
- **Concurrent access patterns**
- **Cache size management**
- **Cold start scenarios**

> NOTE: There is no single correct answer here! Discuss the trade-offs with your group.

##### ✅ Answer:

**Limitations of This Production Caching Approach:**

**Memory vs Disk Trade-offs:**
- Current in-memory cache is fast but volatile - loses all cached data on restart
- 14.4x speedup disappears when service restarts, requiring cache rebuild

**Cache Invalidation:**
- No expiration strategy - stale data persists indefinitely
- Updated PDFs won't invalidate old cached embeddings/responses

**Concurrent Access:**
- Cache stampeding: multiple users requesting same uncached item hit API simultaneously
- No thread-safe protection for concurrent cache operations

**Cache Size Management:**
- No size limits or eviction policies - unbounded memory growth
- Missing LRU/LFU strategies when memory limits reached

**Cold Start Scenarios:**
- Every deployment/restart loses entire cache, eliminating performance benefits
- New instances start with empty cache, causing poor initial performance

**When MOST Useful:**
- Development/testing environments with repeated queries
- Single-user applications with stable content
- FAQ-style apps with repeated identical questions

**When LEAST Useful:**
- Multi-instance production deployments (no shared cache)
- Dynamic content systems with frequent document updates
- High-concurrency applications requiring cache coordination
- 24/7 services where restarts cause performance degradation

The approach demonstrates excellent performance gains but needs persistence and distributed caching for production readiness.

Test Embedding Cache Performance:

````python
#1.# Embedding Cache Performance
import time
import uuid
import shutil
from statistics import mean
from pathlib import Path

from langgraph_agent_lib.caching import CacheBackedEmbeddings, setup_llm_cache
from langgraph_agent_lib.models import get_openai_model

# Ensure your OpenAI key is available
#assert os.getenv("OPENAI_API_KEY"), "Set OPENAI_API_KEY in your environment before running."
Path("./cache").mkdir(parents=True, exist_ok=True)

def time_call(fn, *args, **kwargs):
    start = time.perf_counter()
    result = fn(*args, **kwargs)
    return result, time.perf_counter() - start

def run_embedding_cache_benchmark(
    texts, 
    repeats=5, 
    model="text-embedding-3-small"
):
    test_cache_dir = f"./cache/embeddings_benchmark_{uuid.uuid4().hex[:8]}"
    # Fresh cache for a clean experiment
    if os.path.exists(test_cache_dir):
        shutil.rmtree(test_cache_dir)
    Path(test_cache_dir).mkdir(parents=True, exist_ok=True)

    embeddings = CacheBackedEmbeddings(model=model, cache_dir=test_cache_dir).get_embeddings()

    first_times = []
    # First pass (misses expected)
    for t in texts:
        _, t_sec = time_call(embeddings.embed_query, t)
        first_times.append(t_sec)

    subsequent_times = []
    # Subsequent passes (hits expected)
    for _ in range(repeats - 1):
        for t in texts:
            _, t_sec = time_call(embeddings.embed_query, t)
            subsequent_times.append(t_sec)

    # Count cached items (rough proxy for unique texts)
    file_count = sum(len(files) for _, _, files in os.walk(test_cache_dir))
    unique_texts = len(set(texts))
    total_calls = len(texts) * repeats
    expected_hit_rate = (total_calls - unique_texts) / total_calls

    return {
        "first_mean_s": mean(first_times) if first_times else 0.0,
        "subsequent_mean_s": mean(subsequent_times) if subsequent_times else 0.0,
        "speedup_x": (mean(first_times) / mean(subsequent_times)) if subsequent_times else float("inf"),
        "cache_dir": test_cache_dir,
        "unique_texts": unique_texts,
        "total_calls": total_calls,
        "expected_hit_rate": expected_hit_rate,
        "cached_file_count": file_count,
    }

texts = [
    "What are the 2024 student loan default rates?",
    #changed to a unique question since having the same question twice doesn't make sense since each question is repeated 5 times
    "Can I defer my student loan payments?", 
]
emb_res = run_embedding_cache_benchmark(texts=texts, repeats=10)
emb_res
````

Test LLM Cache Performance

````python
import os
from statistics import mean
from langchain_core.globals import set_llm_cache

from langgraph_agent_lib.rag import ProductionRAGChain
from langgraph_agent_lib.caching import setup_llm_cache

# Build a RAG chain once so we don't include construction time in the benchmark
pdf_path = "data/The_Federal_Pell_Grant_Program.pdf"  # choose any PDF in ./data
rag = ProductionRAGChain(file_path=pdf_path, llm_model=os.environ.get("OPENAI_MODEL", "gpt-4.1-mini"))

def warm_retrieval(chain, question: str):
    # Warm the query embedding and retriever to isolate LLM cache impact
    _ = chain.get_retriever().invoke(question)

def benchmark_rag_llm_cache(chain, question: str, repeats: int = 6):
    # Ensure no LLM cache, but keep embedding cache warm to focus on LLM caching
    set_llm_cache(None)
    warm_retrieval(chain, question)

    baseline_times = []
    for _ in range(repeats):
        _, t_sec = time_call(chain.invoke, question)
        baseline_times.append(t_sec)

    # Now enable in-memory LLM cache and run the same question
    setup_llm_cache("memory")
    cached_times = []
    for _ in range(repeats):
        _, t_sec = time_call(chain.invoke, question)
        cached_times.append(t_sec)

    # First call with cache is a miss, subsequent should be hits
    cached_first = cached_times[0]
    cached_subseq = cached_times[1:] if len(cached_times) > 1 else []

    return {
        "repeats": repeats,
        "baseline_mean_s": mean(baseline_times),
        "cached_first_s": cached_first,
        "cached_subsequent_mean_s": mean(cached_subseq) if cached_subseq else 0.0,
        "speedup_cached_vs_baseline_x": (mean(baseline_times) / (mean(cached_subseq) if cached_subseq else float("inf"))),
    }

def pretty_print_rag_llm(res):
    print("== RAG LLM Cache (no-cache vs memory) ==")
    print(f"Repeats: {res['repeats']}")
    print(f"Baseline mean (no cache): {res['baseline_mean_s']:.3f}s")
    print(f"With cache - first (miss): {res['cached_first_s']:.3f}s")
    print(f"With cache - subsequent mean (hits): {res['cached_subsequent_mean_s']:.3f}s")
    print(f"Speedup (cached hits vs baseline): {res['speedup_cached_vs_baseline_x']:.1f}x")

question = "In one sentence, explain the Pell Grant program."
rag_llm_res = benchmark_rag_llm_cache(rag, question, repeats=6)
pretty_print_rag_llm(rag_llm_res)
````

````python
 ##### ✅ Answer:

from langchain_core.messages import HumanMessage, AIMessage, ToolMessage
import time
### YOUR EXPERIMENTATION CODE HERE ###

# Example: Test different query types
queries_to_test = [
    "What is the main purpose of the Direct Loan Program?",  # RAG-focused
    "What are the latest developments in AI safety?",  # Web search
    "Find recent papers about transformer architectures",  # Academic search
    "How do the concepts in this document relate to current AI research trends?"  # Multi-tool
]


def extract_helpfulness(msgs):
    # Find latest helpfulness message and index
    decision, decision_idx = None, None
    for i in range(len(msgs) - 1, -1, -1):
        text = getattr(msgs[i], "content", "") or ""
        if isinstance(msgs[i], AIMessage) and isinstance(text, str) and text.startswith("HELPFULNESS:"):
            decision = text.split(":", 1)[1].strip()
            decision_idx = i
            break
    # Count all helpfulness checks
    loops = sum(
        1
        for m in msgs
        if isinstance(m, AIMessage)
        and isinstance(getattr(m, "content", ""), str)
        and m.content.startswith("HELPFULNESS:")
    )
    # Final answer is the last AI message before the helpfulness decision
    answer = None
    if decision_idx is not None:
        for j in range(decision_idx - 1, -1, -1):
            if isinstance(msgs[j], AIMessage) and getattr(msgs[j], "content", ""):
                answer = msgs[j].content
                break
    else:
        # Fallback: last AI message
        for j in range(len(msgs) - 1, -1, -1):
            if isinstance(msgs[j], AIMessage) and getattr(msgs[j], "content", ""):
                answer = msgs[j].content
                break
    return decision, answer, loops

def count_agent_tooling(msgs):
    agent_turns = 0
    agent_turns_with_tools = 0
    total_tool_calls = 0
    tool_messages = 0
    for m in msgs:
        if isinstance(m, AIMessage):
            agent_turns += 1
            calls = getattr(m, "tool_calls", None) or []
            if calls:
                agent_turns_with_tools += 1
                total_tool_calls += len(calls)
        elif isinstance(m, ToolMessage):
            tool_messages += 1
    return agent_turns, agent_turns_with_tools, total_tool_calls, tool_messages

for query in queries_to_test:
    print(f"\n🔍 Testing: {query}")

    # Simple agent
    t0 = time.perf_counter()
    simple_resp = simple_agent.invoke({"messages": [HumanMessage(content=query)]})
    simple_s = time.perf_counter() - t0
    simple_answer = next(
        (m.content for m in reversed(simple_resp["messages"])
         if isinstance(m, AIMessage) and getattr(m, "content", "")),
        None,
    )
    print(f"Simple Agent (elapsed: {simple_s:.2f}s)")
    print(simple_answer or "[no AI answer]")

    # Helpfulness agent
    t0 = time.perf_counter()
    helpful_resp = helpfulness_agent.invoke({"messages": [HumanMessage(content=query)]})
    helpful_s = time.perf_counter() - t0
    decision, helpful_answer, helpfulness_loops = extract_helpfulness(helpful_resp["messages"])
    a_turns, a_turns_with_tools, total_tool_calls, tool_msgs = count_agent_tooling(helpful_resp["messages"])
    print(f"\nHelpfulness Agent (elapsed: {helpful_s:.2f}s)")
    print(f"- helpfulness loops: {helpfulness_loops}")
    print(f"- agent turns (all): {a_turns}")
    print(f"- agent turns with tools: {a_turns_with_tools}")
    print(f"- total tool calls: {total_tool_calls}")
    print(f"- tool result messages: {tool_msgs}")
    print(f"- decision: {decision or 'N/A'}")
    print(helpful_answer or "[no AI answer]")

    # Quick compare
    print("\n— Compare —")
    faster = "Simple" if simple_s < helpful_s else "Helpfulness"
    print(f"Latency: {faster} faster ({simple_s:.2f}s vs {helpful_s:.2f}s)")
    if simple_answer and helpful_answer and simple_answer != helpful_answer:
        print("Answers differ.")
````

## Task 4: Guardrails Integration for Production Safety

Now we'll integrate **Guardrails AI** into our production system to ensure our agents operate safely and within acceptable boundaries. Guardrails provide essential safety layers for production LLM applications by validating inputs, outputs, and behaviors.

### 🛡️ What are Guardrails?

Guardrails are specialized validation systems that help "catch" when LLM interactions go outside desired parameters. They operate both **pre-generation** (input validation) and **post-generation** (output validation) to ensure safe, compliant, and on-topic responses.

**Key Categories:**
- **Topic Restriction**: Ensure conversations stay on-topic
- **PII Protection**: Detect and redact sensitive information  
- **Content Moderation**: Filter inappropriate language/content
- **Factuality Checks**: Validate responses against source material
- **Jailbreak Detection**: Prevent adversarial prompt attacks
- **Competitor Monitoring**: Avoid mentioning competitors

### Production Benefits of Guardrails

**🏢 Enterprise Requirements:**
- **Compliance**: Meet regulatory requirements for data protection
- **Brand Safety**: Maintain consistent, appropriate communication tone
- **Risk Mitigation**: Reduce liability from inappropriate AI responses
- **Quality Assurance**: Ensure factual accuracy and relevance

**⚡ Technical Advantages:**
- **Layered Defense**: Multiple validation stages for robust protection
- **Selective Enforcement**: Different guards for different use cases
- **Performance Optimization**: Fast validation without sacrificing accuracy
- **Integration Ready**: Works seamlessly with LangGraph agent workflows

````python
if guardrails_available:
    print("🛡️ Setting up production Guardrails...")
    
    # 1. Topic Restriction Guard - Keep conversations focused on student loans
    topic_guard = Guard().use(
        RestrictToTopic(
            valid_topics=["student loans", "financial aid", "education financing", "loan repayment"],
            invalid_topics=["investment advice", "crypto", "gambling", "politics"],
            disable_classifier=True,
            disable_llm=False,
            on_fail="exception"
        )
    )
    print("✓ Topic restriction guard configured")
    
    # 2. Jailbreak Detection Guard - Prevent adversarial attacks
    jailbreak_guard = Guard().use(DetectJailbreak())
    print("✓ Jailbreak detection guard configured")
    
    # 3. PII Protection Guard - Protect sensitive information
    pii_guard = Guard().use(
        GuardrailsPII(
            entities=["CREDIT_CARD", "SSN", "PHONE_NUMBER", "EMAIL_ADDRESS"], 
            on_fail="fix"
        )
    )
    print("✓ PII protection guard configured")
    
    # 4. Content Moderation Guard - Keep responses professional
    profanity_guard = Guard().use(
        ProfanityFree(threshold=0.8, validation_method="sentence", on_fail="exception")
    )
    print("✓ Content moderation guard configured")
    
    # 5. Factuality Guard - Ensure responses align with context
    factuality_guard = Guard().use(
        LlmRagEvaluator(
            eval_llm_prompt_generator=HallucinationPrompt(prompt_name="hallucination_judge_llm"),
            llm_evaluator_fail_response="hallucinated",
            llm_evaluator_pass_response="factual", 
            llm_callable="gpt-4.1-mini",
            on_fail="exception",
            on="prompt"
        )
    )
    print("✓ Factuality guard configured")
    
    print("\\n🎯 All Guardrails configured for production use!")
    
else:
    print("⚠ Skipping Guardrails setup - not available")
````
